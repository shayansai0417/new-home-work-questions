```python
weight = float(input())
height_cm = float(input())

height_m = height_cm / 100
bmi = round(weight / (height_m ** 2), 1)

underweight = bmi < 18.5
normal = 18.5 <= bmi < 25.0
overweight = bmi >= 25.0

print(f"BMI: {bmi}")
print(f"Underweight: {underweight}")
print(f"Normal: {normal}")
print(f"Overweight: {overweight}")
```
